```python
#from os import write
from manim_slides import Slide, ThreeDSlide
from manim_slides.slide import MANIM, MANIMGL

if MANIM:
    from manim import *
elif MANIMGL:
    from manimlib import *
"""WARNING: this example does not seem to work with ManimGL."""
```

```python
%%manim_slides -v WARNING --progress_bar None IntroductoryPart --manim-slides controls=true data_uri=true

class IntroductoryPart(Slide):
    def construct(self):
        self.wait_time_between_slides = 0.1

        pack = ImageMobject("images/pallet_2_view.png")
        pack.scale(0.5)
        pack.shift(2*UP + 5*LEFT)
        self.add((pack))
 
        vdb = ImageMobject("images/vbdimage.png")
        vdb.scale(0.5)
        vdb.shift(2*UP + 5*RIGHT)
        self.add((vdb))

        title = VGroup(
            Text("Упаковка товару.  ", t2c={"Упаковка": BLUE}),
            Text("Векторна база даних.", t2c={"Векторна база": BLUE}),
            Text("Що спільного між ними?", t2w={"[-22:]": BOLD}, t2c={"[-22:]": YELLOW}),
        ).arrange(DOWN)

        self.play(FadeIn(title))

        self.next_slide()
        self.play(FadeOut(title))

        docs = ImageMobject("images/Screenshot.jpg")
        docs.scale(0.5)
        self.add((docs))
        self.play(docs.animate.shift(2*LEFT+UP*2).scale(0.5))
        narration_text = Text("Упаковка товару – це процес.", font_size=30)
        narration_text.next_to(docs, DOWN, buff=1)
        self.play(FadeIn(narration_text))

        self.wait(3)
        #self.play(FadeOut(docs,narration_text))

        vbd = ImageMobject("images/lesmis.jpg")
        vbd.scale(0.5)
        self.add((vbd))
        self.play(vbd.animate.shift(2*RIGHT+UP*2).scale(0.4))
        narration_text = Text("Векторна база даних - ... це ми розглянемо далі", font_size=30)
        narration_text.next_to(narration_text, DOWN, buff=0.5)
        self.play(FadeIn(narration_text))

        
        self.next_slide()
        #self.play(FadeOut(vbd,narration_text))
```
